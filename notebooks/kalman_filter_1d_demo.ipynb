```python
%load_ext autoreload
%autoreload 2

import random

import matplotlib.pyplot as plt

from math_in_python.numerical.kalman_filter_1d import kalman_filter_1d
```

```python
rng = random.Random(0)
true_value = 10.0
zs = [true_value + rng.gauss(0, 2) for _ in range(50)]
est = kalman_filter_1d(zs, x0=0.0, P0=1000.0, Q=1e-4, R=4.0)

plt.plot(zs, ".", label="measurements")
plt.plot(est, label="Kalman estimate")
plt.axhline(true_value, color="gray", linestyle="--", label="true value")
plt.xlabel("step")
plt.legend()
plt.show()
```
